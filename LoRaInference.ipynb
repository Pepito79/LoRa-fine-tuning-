```python
from transformers import AutoModelForCausalLM, AutoTokenizer ,pipeline 
import torch
from pprint import pprint
import textwrap
```

```python
d = {"name":"sadd","age":18}
def func(name,age):
    print(name,age)
func(**d)
```

```python
final_model = "./merged-model"

model = AutoModelForCausalLM.from_pretrained(final_model)
tokenizer = AutoTokenizer.from_pretrained(final_model)
```

```python
pipe = pipeline('text-generation', model=model,tokenizer= tokenizer)
```

```python
len_wrap = 70
def generate(text):
    instruction = ("<|system|>"
                    "Please answer with one and only one of the option in the bracket" 
                    "That have been given by the user"
                    "And explain why your choice."
                    "</s")
    tex = ("<|user|>" "Q : "+text+"</s>")
    tex = instruction + tex
    answer = pipe(tex, max_new_tokens=150, top_k=50, temperature=0.7, return_full_text=False)
    if not answer or answer[0]["generated_text"] =='':
        print("Bug detected please rerun ")
    l= textwrap.wrap(answer[0]['generated_text'].replace("'",""),len_wrap)
    final_ans = "\n".join(l)
    print(final_ans)
```

```python
text = ("Patient is a 55-year-old male smoker with a new, severe, right-sided "
        "headache (7/10) that started 3 days ago. Given these risk factors and "
        "symptoms, the most concerning diagnosis is one of the following:"
        " { A) :Migraine without aura B) Temporomandibular joint disorder (TMJ) "
        "C) Temporal Arteritis (Giant cell arteritis)}. Choose the best answer. "
        )
generate(text)
```
